```python
import json
import pandas as pd

from configuration.configuration import EncoderConfig, Configuration
from preprocessing_kit.preprocessing.data_processor import DataProcessor
from preprocessing_kit.preprocessing.processor import Processor

%load_ext autoreload
%autoreload 2
```

```python
df_init = pd.read_csv("../data/bank-additional-full.csv", sep=';')
df = df_init.copy()
df.head(3)
```

```python
with open("../config.json") as f:
    raw = json.load(f)

configuration = Configuration(
    encoding_config=EncoderConfig(**raw["encoder"]),
    outlier_config=raw["outlier_config"],
    missing_values=raw["missing_values"],
    feature_engineering=raw["feature_engineering"]
)
```

```python
processor = Processor(target_col="y", unnecessary_columns=["duration"], scaling_method="standard", config=configuration)
data_processor = DataProcessor(
    processor=processor,
    apply_outliers_processing=True,
    apply_feature_engineering=True,
    config=configuration
)

X_train, y_train, X_val, y_val = data_processor.process(df)
```
